```python
# The raw input string
raw_data = "  \t  joHN_doe_1995!! || data_scientist || python,SQL,machine_learning || john.doe@email.com || $120,000  \n"

# ==========================================
# 1. STRIPPING & CLEANING (Whitespace removal)
# ==========================================
# Remove leading/trailing whitespaces, tabs, and newlines
cleaned_data = raw_data.strip()
print(cleaned_data)
# ('lstrip' and 'rstrip' could be used if you only wanted to clean one side)

# ==========================================
# 2. SPLITTING & PARTITIONING
# ==========================================
# Split the big string into its core components based on the delimiter
components = cleaned_data.split(" || ")
print(components)
raw_username, role, skills_str, email, salary_str = components

# Partition can be used if we just want to isolate the username from the rest instantly
first_part, separator, rest = cleaned_data.partition(" || ")

# ==========================================
# 3. CASE TRANSFORMATIONS
# ==========================================
# Let's fix the case of the role
role_upper = role.upper()          # "DATA_SCIENTIST"
role_lower = role.lower()          # "data_scientist"
role_title = role.title()          # "Data_Scientist"
role_cap   = role.capitalize()     # "Data_scientist"
role_swap  = role_upper.swapcase() # "data_scientist"

# ==========================================
# 4. REPLACING & ALTERING
# ==========================================
# Clean up the role name for display
clean_role = role_title.replace("_", " ") # "Data Scientist"

# Clean up salary to make it a pure numeric string later
clean_salary = salary_str.replace("$", "").replace(",", "") # "120000"

# ==========================================
# 5. SEARCHING & FINDING (Locating Substrings)
# ==========================================
# Find where the domain starts in the email
domain_index = email.find("@")   # Returns index of '@'
r_domain_idx = email.rfind(".")  # Finds the last dot (right-most search)

# index() and rindex() do the same but raise a ValueError if not found
try:
    dot_index = email.index(".")
except ValueError:
    pass

# Count how many skills the user listed
skill_count = skills_str.count(",") + 1

# ==========================================
# 6. VALIDATION & CHARACTER CHECKS
# ==========================================
# Check username properties after stripping special characters manually or via a check
username_clean = raw_username.rstrip("!").strip() # "joHN_doe_1995"

# Run boolean checks across various components
is_alnum   = username_clean.replace("_", "").isalnum() # True (letters & numbers)
is_alpha   = clean_role.replace(" ", "").isalpha()     # True (only letters)
is_digit   = clean_salary.isdigit()                    # True (only digits)
is_numeric = clean_salary.isnumeric()                  # True
is_decimal = clean_salary.isdecimal()                  # True

# Case checks
is_low   = role_lower.islower()   # True
is_up    = role_upper.isupper()   # True
is_space = "   ".isspace()        # True
is_title = role_title.istitle()   # True
is_ident = "variable_name".isidentifier() # True (checks if it's a valid Python variable name)
is_print = raw_data.isprintable() # False (because it contains \t and \n)

# Prefix / Suffix checks
has_prefix = email.startswith("john")
has_suffix = email.endswith(".com")

# ==========================================
# 7. ALIGNMENT & PADDING (Formatting for tables)
# ==========================================
# Center the role name for a text-based UI dashboard
centered_role = clean_role.center(20, "-") # "---Data Scientist---"
left_aligned  = clean_role.ljust(20, " ")  # "Data Scientist     "
right_aligned = clean_role.rjust(20, " ")  # "     Data Scientist"

# Pad an ID number or salary with leading zeros
padded_salary = clean_salary.zfill(8)      # "00120000"

# ==========================================
# 8. JOINING & ADVANCED FORMATTING
# ==========================================
# Split individual skills, modify them, and rejoin them with a new delimiter
skills_list = skills_str.split(",")
formatted_skills = [s.capitalize() for s in skills_list]
final_skills = ", ".join(formatted_skills) # "Python, Sql, Machine_learning"

# Use mapping for translation (e.g., masking vowels in an email for privacy)
vowel_mask = str.maketrans("aeiou", "*****")
masked_email = email.translate(vowel_mask) # "j*hn.d**@*m**l.c*m"

# Final output string construction using format methods
summary_template = "User {} ({}) specializes in: {}. Earns: ${}."
final_summary = summary_template.format(username_clean, clean_role, final_skills, clean_salary)

# Or using the older mapping format (format_map)
data_dict = {'user': username_clean, 'salary': clean_salary}
map_summary = "Audit log: {user} verified. Base: {salary}".format_map(data_dict)

# encode/decode strings (e.g., for saving to a file or network stream)
byte_data = final_summary.encode('utf-8')
decoded_string = byte_data.decode('utf-8')
```
